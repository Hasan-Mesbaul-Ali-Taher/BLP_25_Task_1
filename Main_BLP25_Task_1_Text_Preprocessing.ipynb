# Text Preprocessing

## Number removal (NR)

```python
import re

# Function to remove numbers
def remove_numbers(text):
    return re.sub(r'\d+', '', text)

# Apply number removal to all relevant text columns
train_1A_df['nr_text'] = train_1A_df['text'].apply(remove_numbers)
dev_1A_df['nr_text'] = dev_1A_df['text'].apply(remove_numbers)
dev_test_1A_df['nr_text'] = dev_test_1A_df['text'].apply(remove_numbers)
test_1A_df['nr_text'] = test_1A_df['text'].apply(remove_numbers)

train_1B_df['nr_text'] = train_1B_df['text'].apply(remove_numbers)
dev_1B_df['nr_text'] = dev_1B_df['text'].apply(remove_numbers)
dev_test_1B_df['nr_text'] = dev_test_1B_df['text'].apply(remove_numbers)
test_1B_df['nr_text'] = test_1B_df['text'].apply(remove_numbers)

train_1C_df['nr_text'] = train_1C_df['text'].apply(remove_numbers)
dev_1C_df['nr_text'] = dev_1C_df['text'].apply(remove_numbers)
dev_test_1C_df['nr_text'] = dev_test_1C_df['text'].apply(remove_numbers)
test_1C_df['nr_text'] = test_1C_df['text'].apply(remove_numbers)

print("Numbers removed from text columns in all datasets.")
```

## Tokenization

```python
# Function for tokenization
def tokenize_text(text):
    return text.split()

# Apply tokenization to all relevant text columns
train_1A_df['t_text'] = train_1A_df['nr_text'].apply(tokenize_text)
dev_1A_df['t_text'] = dev_1A_df['nr_text'].apply(tokenize_text)
dev_test_1A_df['t_text'] = dev_test_1A_df['nr_text'].apply(tokenize_text)
test_1A_df['t_text'] = test_1A_df['nr_text'].apply(tokenize_text)

train_1B_df['t_text'] = train_1B_df['nr_text'].apply(tokenize_text)
dev_1B_df['t_text'] = dev_1B_df['nr_text'].apply(tokenize_text)
dev_test_1B_df['t_text'] = dev_test_1B_df['nr_text'].apply(tokenize_text)
test_1B_df['t_text'] = test_1B_df['nr_text'].apply(tokenize_text)

train_1C_df['t_text'] = train_1C_df['nr_text'].apply(tokenize_text)
dev_1C_df['t_text'] = dev_1C_df['nr_text'].apply(tokenize_text)
dev_test_1C_df['t_text'] = dev_test_1C_df['nr_text'].apply(tokenize_text)
test_1C_df['t_text'] = test_1C_df['nr_text'].apply(tokenize_text)

print("Text columns tokenized in all datasets.")
```

```python
from pandarallel import pandarallel
from tqdm import tqdm

# Initialize pandarallel
pandarallel.initialize()

# Combine the lemma datasets
combined_lemma_df = pd.concat([bangla_lem_df, banel_df]).drop_duplicates().set_index('Inflected_Word')

# Create a dictionary for faster lookups
lemma_dict = combined_lemma_df['Lemma'].to_dict()

# Function for lemmatization
def lemmatize_text(tokens):
    return [lemma_dict.get(token, token) for token in tokens]

# Apply lemmatization using pandarallel
train_1A_df['l_nr_text'] = train_1A_df['t_text'].parallel_apply(lemmatize_text)
dev_1A_df['l_nr_text'] = dev_1A_df['t_text'].parallel_apply(lemmatize_text)
dev_test_1A_df['l_nr_text'] = dev_test_1A_df['t_text'].parallel_apply(lemmatize_text)
test_1A_df['l_nr_text'] = test_1A_df['t_text'].parallel_apply(lemmatize_text)

train_1B_df['l_nr_text'] = train_1B_df['t_text'].parallel_apply(lemmatize_text)
dev_1B_df['l_nr_text'] = dev_1B_df['t_text'].parallel_apply(lemmatize_text)
dev_test_1B_df['l_nr_text'] = dev_test_1B_df['t_text'].parallel_apply(lemmatize_text)
test_1B_df['l_nr_text'] = test_1B_df['t_text'].parallel_apply(lemmatize_text)

train_1C_df['l_nr_text'] = train_1C_df['t_text'].parallel_apply(lemmatize_text)
dev_1C_df['l_nr_text'] = dev_1C_df['t_text'].parallel_apply(lemmatize_text)
dev_test_1C_df['l_nr_text'] = dev_test_1C_df['t_text'].parallel_apply(lemmatize_text)
test_1C_df['l_nr_text'] = test_1C_df['t_text'].parallel_apply(lemmatize_text)

print("Text columns lemmatized in all datasets.")
```
