```python
import pandas as pd
```

# Load all the datasets

## Task 1A

```python
github_link = "https://github.com/Hasan-Mesbaul-Ali-Taher/Temp/blob/master/"
raw_link = github_link.replace("blob/", "").replace("github.com", "raw.githubusercontent.com")

train_1A_df = pd.read_csv(raw_link+"blp25_hatespeech_subtask_1A_train.tsv", sep="\t", keep_default_na=False)
dev_1A_df = pd.read_csv(raw_link+"blp25_hatespeech_subtask_1A_dev.tsv", sep="\t", keep_default_na=False)
dev_test_1A_df = pd.read_csv(raw_link+"blp25_hatespeech_subtask_1A_dev_test.tsv", sep="\t", keep_default_na=False)
test_1A_df = pd.read_csv(raw_link+"blp25_hatespeech_subtask_1A_test.tsv", sep="\t", keep_default_na=False)
```

```python
train_1A_df
```

# Simple Classification

## TF-IDF Vectorizer

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    tokenizer=lambda x: x.split(),
    ngram_range=(1,1),  # for trigram use (1,3), Bi-gram use (1,2), Unigram use (1,1)
    min_df=3,
    sublinear_tf=True
)
# Fit the vectorizer on the concatenated text data
vectorizer.fit(pd.concat([train_1A_df['sr_l_nr_text'], dev_1A_df['sr_l_nr_text'], dev_test_1A_df['sr_l_nr_text'], test_1A_df['sr_l_nr_text']], ignore_index=True).fillna(''))

X_train_vectorized = vectorizer.transform(train_1A_df['sr_l_nr_text'])
X_dev_vectorized = vectorizer.transform(dev_1A_df['sr_l_nr_text'])
X_dev_test_vectorized = vectorizer.transform(dev_test_1A_df['sr_l_nr_text'])
X_test_vectorized = vectorizer.transform(test_1A_df['sr_l_nr_text'])

print("Shape of vectorized train data:", X_train_vectorized.shape)
print("Shape of vectorized dev data:", X_dev_vectorized.shape)
print("Shape of vectorized dev_test data:", X_dev_test_vectorized.shape)
print("Shape of vectorized test data:", X_test_vectorized.shape)
```

```python
print(vectorizer.get_feature_names_out())
```

## Prepare data for modeling


```python
y_train = train_1A_df['label']
y_dev = dev_1A_df['label']
y_dev_test = dev_test_1A_df['label']
y_test = test_1A_df['label']
print("Shape of y_train:", y_train.shape)
print("Shape of y_dev:", y_dev.shape)
print("Shape of y_dev_test:", y_dev_test.shape)
print("Shape of y_test:", y_test.shape)
```

## Choose a model



```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(random_state=1804038, max_iter=1000)
```

## Train and Evaluate the Baseline model



```python
from sklearn.metrics import f1_score

model.fit(X_train_vectorized, y_train)

baseline_preds_train = model.predict(X_train_vectorized)
baseline_preds_dev = model.predict(X_dev_vectorized)
baseline_preds_dev_test = model.predict(X_dev_test_vectorized)
baseline_preds_test = model.predict(X_test_vectorized)

print("Baseline Train F1 Score:", f1_score(y_train, baseline_preds_train, average='micro'))
print("Baseline Dev F1 Score:", f1_score(y_dev, baseline_preds_dev, average='micro'))
print("Baseline Dev_test F1 Score:", f1_score(y_dev, baseline_preds_dev, average='micro'))
print("Baseline Test F1 Score:", f1_score(y_test, baseline_preds_test, average='micro'))
```

```python
model = LogisticRegression(random_state=1804038, max_iter=1000, C=1, class_weight=None, solver='liblinear')
```

## Train the Hyperparameter Tuned Model on Full training data + Dev data


```python
from scipy.sparse import vstack

# Concatenate the vectorized features vertically
X_train_dev_vectorized = vstack([X_train_vectorized, X_dev_vectorized])

# Concatenate the labels vertically
y_train_dev = pd.concat([y_train, y_dev], ignore_index=True)

# Train the model on the combined data
model.fit(X_train_dev_vectorized, y_train_dev)
```

# Re-evaluate model

```python
from sklearn.metrics import classification_report, accuracy_score

# Make predictions on the train set using the best model
y_train_pred = model.predict(X_train_vectorized)

# Evaluate the model's performance on the train set
print("Classification Report on Train Set:")
print(classification_report(y_train, y_train_pred))

print("Accuracy on Train Set:", accuracy_score(y_train, y_train_pred))

print("-" * 50) # Separator for clarity

# Make predictions on the dev set using the best model
y_dev_pred = model.predict(X_dev_vectorized)

# Evaluate the model's performance on the dev set
print("Classification Report on Dev Set:")
print(classification_report(y_dev, y_dev_pred))

print("Accuracy on Dev Set:", accuracy_score(y_dev, y_dev_pred))

print("-" * 50) # Separator for clarity

# Make predictions on the dev set using the best model
y_dev_test_pred = model.predict(X_dev_test_vectorized)

# Evaluate the model's performance on the dev set
print("Classification Report on Dev Test Set:")
print(classification_report(y_dev_test, y_dev_test_pred))

print("Accuracy on Dev Test Set:", accuracy_score(y_dev_test, y_dev_test_pred))

print("-" * 50) # Separator for clarity

# Make predictions on the dev set using the best model
y_test_pred = model.predict(X_test_vectorized)

# Evaluate the model's performance on the dev set
print("Classification Report on Test Set:")
print(classification_report(y_test, y_test_pred))

print("Accuracy on Test Set:", accuracy_score(y_test, y_test_pred))

```

## Visualize Predictions vs. Real Labels
